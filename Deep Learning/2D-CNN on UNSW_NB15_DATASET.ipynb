```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
```

```python
# Load the dataset
data = pd.read_csv('imbalance_data_UNSW_NB15_Dataset.csv')
```

```python
# Handle infinite values and large values in the features
data.replace([np.inf, -np.inf], np.nan, inplace=True)
data.fillna(data.mean(), inplace=True)
```

```python
# Split the data into features and labels
X = data.drop(columns=['label'])
y = data['label']
```

```python
# Determine the size of the square matrix
num_features = X.shape[1]
square_size = int(np.ceil(np.sqrt(num_features)))
```

```python
# Pad the features to fit into a square matrix
padding_size = square_size**2 - num_features
X_padded = np.pad(X.values, ((0, 0), (0, padding_size)), 'constant')
```

```python
# Reshape the data into 2D format (square_size x square_size)
X_reshaped = X_padded.reshape(-1, square_size, square_size, 1)
```

```python
# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_reshaped, y, test_size=0.2, random_state=42)
```

```python
# Define the 2D-CNN model
model = Sequential()
model.add(Conv2D(filters=32, kernel_size=(3, 3), activation='relu', input_shape=(square_size, square_size, 1)))
model.add(MaxPooling2D(pool_size=(2, 2)))
model.add(Flatten())
model.add(Dense(128, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
```

```python
# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# Train the model
model.fit(X_train, y_train, epochs=10, batch_size=32, validation_data=(X_test, y_test))

# Evaluate the model on the test set
loss, accuracy = model.evaluate(X_test, y_test)
print(f'Test Accuracy: {accuracy}')
```
